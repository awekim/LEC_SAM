```python
import pandas as pd
hack_sal = pd.read_csv('/content/drive/MyDrive/[YU]/[Lecture]/[SAM]/01_SAM_Practice/SAM_LEC04/salaries_clean_edited.csv')
hack_sal.head()
```

```python
# P1. Which job category pays the most on average?
# groupby + mean
avg_pay_by_category = hack_sal.groupby('job_title_category')['annual_base_pay'].mean().reset_index()
avg_pay_by_category
```

```python
highest_paying = avg_pay_by_category.sort_values(by='annual_base_pay', ascending=False)
print(highest_paying)
```

```python
# P2. What kind of job do most of the people in this survey do?
# groupby + size/count
people_count_by_category = hack_sal.groupby('job_title_category').size().reset_index(name='count')
people_count_by_category
```

```python
most_people = people_count_by_category.sort_values(by='count', ascending=False)
print(most_people)
```

```python
# P3. Break down the average salary by both job category and rank to show me the detailed differences
# 2차원 groupby + mean
detailed_salary = hack_sal.groupby(['job_title_category', 'job_title_rank'])['annual_base_pay'].mean().reset_index()
print(detailed_salary)
```

```python
# P4. The salary numbers are quite large and hard to compare at a glance.
# Convert the annual_base_pay into a scale from 0 to 1, so we can easily see
# where each person stands between the absolute minimum and maximum salaries.

# Min-Max Normalization
min_pay = hack_sal['annual_base_pay'].min()
max_pay = hack_sal['annual_base_pay'].max()

hack_sal['normalized_pay'] = (hack_sal['annual_base_pay'] - min_pay) / (max_pay - min_pay)

# def min_max_normalize(x):
#     return (x - x.min()) / (x.max() - x.min())
# hack_sal['normalized_pay'] = min_max_normalize(hack_sal['annual_base_pay'])

print(hack_sal[['job_title', 'annual_base_pay', 'normalized_pay']].head())
```

```python
# z-score normalization
def standardization(x):
    return (x - x.mean()) / x.std()
# standardization = lambda x: (x - x.mean()) / x.std()

hack_sal['standardized_pay'] = standardization(hack_sal['annual_base_pay'])

print(hack_sal[['job_title', 'annual_base_pay', 'standardized_pay']].head())
```

```python
import pandas as pd
import numpy as np
import random

np.random.seed(42)
random.seed(42)

# 1. k_content_master.csv (총 200개 작품)
# 카테고리: 웹툰, 드라마/영화, 예능, 게임
categories = ['Webtoon', 'Drama/Movie', 'Variety Show', 'Game']
content_ids = [f'K{str(i).zfill(3)}' for i in range(1, 201)]

master_data = {
    'Content_ID': content_ids,
    'Title': [f'Content_Title_{i}' for i in range(1, 201)],
    'Category': np.random.choice(categories, 200, p=[0.4, 0.3, 0.1, 0.2]),
    'Release_Year': np.random.choice(range(2020, 2027), 200)
}
pd.DataFrame(master_data).to_csv('/content/drive/MyDrive/[YU]/[Lecture]/[SAM]/01_SAM_Practice/SAM_LEC04/k_content_master.csv', index=False)

# 2. platform_distribution.csv (1:N 관계 포함, 한 작품이 여러 플랫폼에 유통됨)
platforms_pool = {
    'Webtoon': ['Naver Webtoon', 'KakaoPage', 'Lezhin'],
    'Drama/Movie': ['Netflix', 'Tving', 'Wavve', 'Disney+'],
    'Variety Show': ['Netflix', 'Tving', 'SBS', 'tvN'],
    'Game': ['Steam', 'Nexon', 'Google Play', 'Kakao Games']
}
monetization_pool = ['SVOD', 'AVOD', 'Freemium', 'Premium']

dist_records = []
for i in range(200):
    c_id = master_data['Content_ID'][i]
    cat = master_data['Category'][i]

    # 작품당 1~3개의 플랫폼에 유통 (멀티 호밍)
    num_platforms = random.choices([1, 2, 3], weights=[0.6, 0.3, 0.1])[0]
    selected_platforms = random.sample(platforms_pool[cat], num_platforms)

    for p in selected_platforms:
        is_exclusive = 'Y' if num_platforms == 1 else 'N'
        dist_records.append({
            'Content_ID': c_id,
            'Platform': p,
            'Monetization': random.choice(monetization_pool),
            'Exclusive_YN': is_exclusive
        })
pd.DataFrame(dist_records).to_csv('/content/drive/MyDrive/[YU]/[Lecture]/[SAM]/01_SAM_Practice/SAM_LEC04/platform_distribution.csv', index=False)

# 3. engagement_metrics.csv (결측치 포함)
metrics_records = []
dist_df = pd.DataFrame(dist_records)

for _, row in dist_df.iterrows():
    # 2026년 신작 중 일부는 아직 성과 데이터가 없도록 처리 (NaN)
    is_new_and_missing = random.random() < 0.1 # 10% 확률로 결측치

    metrics_records.append({
        'Content_ID': row['Content_ID'],
        'Platform': row['Platform'],
        'MAU_Millions': np.nan if is_new_and_missing else round(np.random.lognormal(mean=1.5, sigma=1.0), 2),
        'Revenue_USD_M': np.nan if is_new_and_missing else round(np.random.exponential(scale=50), 2),
        'User_Rating': np.nan if is_new_and_missing else round(np.random.uniform(2.0, 5.0), 1)
    })
pd.DataFrame(metrics_records).to_csv('/content/drive/MyDrive/[YU]/[Lecture]/[SAM]/01_SAM_Practice/SAM_LEC04/engagement_metrics.csv', index=False)

print("고급 실습용 대규모 K-Content 생태계 데이터가 생성되었습니다!")
```

```python
# P2. Does distributing a single title across multiple platforms actually increase
# our bottom line? Compare the average total revenue of content
# distributed on two or more platforms versus content restricted to just one.
# (Hint: Make sure to calculate the total revenue per content first before taking the overall average)
platform_counts = dist.groupby('Content_ID')['Platform'].count().reset_index(name='Plat_Count')
platform_counts['Is_Multi'] = platform_counts['Plat_Count'] > 1
print(platform_counts.head(1))

revenue_by_content = metrics.groupby('Content_ID')['Revenue_USD_M'].sum().reset_index()
print(revenue_by_content.head(1))

q2_df = platform_counts.merge(revenue_by_content, on='Content_ID')
print(q2_df.head(1))

q2_result = q2_df.groupby('Is_Multi')['Revenue_USD_M'].mean().reset_index()
q2_result
```

```python
# P3. What are the top 5 'Content-Platform' combinations with the highest MAU
# specifically within the 'SVOD' (Subscription) model?
q3_merged = dist.merge(metrics, on=['Content_ID', 'Platform'])
print(q3_merged.head(1))

svod_top5 = q3_merged[q3_merged['Monetization'] == 'SVOD'].sort_values('MAU_Millions', ascending=False).head(5)
print(svod_top5.head(1))

svod_top5 = svod_top5.merge(master[['Content_ID', 'Title']], on='Content_ID')
svod_top5
```
