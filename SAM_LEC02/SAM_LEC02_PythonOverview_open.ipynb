```python
def calculate_course_score(midterm, final):
    course_score = midterm * 0.4 + final * 0.6
    return course_score
```

```python
calculate_course_score(68, 82)
```

```python
calculate_course_score(75, 70)
```

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
```

```python
factorial(4)
```

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

number = 5
print(f"The factorial of {number} is {factorial(number)}")
```
