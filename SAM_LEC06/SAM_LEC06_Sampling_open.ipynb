Q1. What is the difference between a target population and a sampling frame?


The target population is the entire group that researchers want to study and draw conclusions about. The sampling frame is the list or source from which the sample is actually selected.
For example, if researchers want to study all university students, the target population includes every university student, while the sampling frame might include only students registered in the university's online system. If some students are excluded from the sampling frame, the sample may not fully represent the target population.

Q2. Why does a large sample not automatically guarantee a valid conclusion?


A large sample reduces random sampling error but does not necessarily eliminate sampling bias. If certain groups are systematically excluded or overrepresented, the results may be biased regardless of sample size.
For example, an online survey with 100,000 respondents may still produce misleading conclusions if students without reliable internet access cannot participate.
Therefore, representativeness matters more than sample size alone.

Q3. How does the law of large numbers differ from the central limit theorem?

The Law of Large Numbers (LLN) states that as the sample size increases, the sample mean tends to approach the population mean.
The Central Limit Theorem (CLT) states that as the sample size increases, the sampling distribution of the sample mean becomes approximately normal, even when the original population is not normally distributed, under appropriate conditions.
In short:
- LLN: Explains why the sample mean approaches the population mean.
- CLT: Explains why the distribution of sample means becomes approximately normal.
Both help explain why large samples are useful in statistical analysis, but they describe different properties of sampling.

Q4. What kinds of students might be missing from an online survey about internet access and academic performance?

Students with limited or unreliable internet access may be underrepresented in an online survey.
For example, students from low-income households, students living in areas with poor internet infrastructure, or students without personal computers may be less likely to participate.
This creates coverage bias, because students who experience the greatest difficulties with internet access may be excluded from the survey.
As a result, the survey may overestimate students' internet access and produce misleading conclusions about its relationship with academic performance.

Q5. What should researchers check before making claims about a population from a sample?

Researchers should evaluate whether the sample adequately represents the target population.
They should consider:
1. Sampling frame: Does the sampling frame adequately cover the target population?
2. Sampling method: How were participants selected? Did the selection process introduce bias?
3. Sample size: Is the sample sufficiently large to provide reasonably precise estimates?
4. Representativeness: Are important population groups appropriately represented?
5. Nonresponse bias: Are certain types of individuals less likely to participate?
Researchers should also consider whether the data were measured reliably.
A large sample alone does not guarantee valid population-level conclusions. The quality of the sampling process is equally important.

This is a coverage or selection bias problem, not merely a small-sample problem.

```python
# P2. The research team can survey only 1,000 respondents.
# Every respondent in the dataset should have the same opportunity to be selected.
# Construct an appropriate sample and estimate the average level of trust in parliament.
sample_1000 = ess.sample(n=1000, replace=False, random_state=123)
sample_mean = sample_1000['trstprl'].mean()

pd.DataFrame({'Value': [population_mean, sample_mean, sample_mean - population_mean]},
             index=['Population mean', 'Sample estimate', 'Sampling error']).round(3)
```

**Interpretation.** A simple random sample is appropriate because every respondent has an equal probability of selection. The sample estimate will usually be close to, but not exactly equal to, the population mean. Their difference is the sampling error observed in this particular sample.

```python
# P3
target_sample_size = 1000
sampling_fraction = target_sample_size / len(ess)

plan_a = (
    ess.groupby('cntry', group_keys=False)
       .sample(frac=sampling_fraction, replace=False, random_state=123)
)

number_of_countries = ess['cntry'].nunique()
n_per_country = target_sample_size // number_of_countries
plan_b = (
    ess.groupby('cntry', group_keys=False)
       .sample(n=n_per_country, replace=False, random_state=123)
)

population_share = ess['cntry'].value_counts(normalize=True).sort_index()
plan_a_share = plan_a['cntry'].value_counts(normalize=True).sort_index()
plan_b_share = plan_b['cntry'].value_counts(normalize=True).sort_index()

composition_q4 = pd.concat(
    [population_share.rename('Population share'),
     plan_a_share.rename('Plan A share'),
     plan_b_share.rename('Plan B share')],
    axis=1
)

print(f'Plan A sample size: {len(plan_a):,}')
print(f'Plan B sample size: {len(plan_b):,}')
print(f'Plan A overall mean: {plan_a["trstprl"].mean():.3f}')
print(f'Plan B overall mean: {plan_b["trstprl"].mean():.3f}')
print(f'Population mean: {population_mean:.3f}')
composition_q4.round(4)
```

**Interpretation.** Plan A is proportionate stratified random sampling. It is more appropriate for estimating overall trust because it preserves the original country composition. Plan B is disproportionate stratified random sampling. It is more appropriate for comparing countries because every country contributes approximately the same number of respondents. However, Plan B's unweighted overall mean gives small and large countries similar influence and therefore does not directly represent the respondent composition of the full dataset.

```python
# P4
countries = ess['cntry'].unique()
np.random.seed(123)
selected_countries = np.random.choice(countries, size=5, replace=False)
cluster_sample = ess[ess['cntry'].isin(selected_countries)]
cluster_mean = cluster_sample['trstprl'].mean()

print('Selected countries:', sorted(selected_countries))
print(f'Cluster sample size: {len(cluster_sample):,}')
print(f'Cluster sample mean: {cluster_mean:.3f}')
print(f'Population mean: {population_mean:.3f}')
print(f'Difference: {cluster_mean - population_mean:.3f}')
```

**Interpretation.** Countries are naturally occurring groups, so randomly selecting countries and observing all respondents within them is cluster sampling. It can reduce fieldwork cost because the research team operates in only a few locations. However, respondents within the same country may be relatively similar, while country means may differ. As a result, the estimate can change substantially depending on which five countries are selected, even when the resulting sample contains many respondents.
