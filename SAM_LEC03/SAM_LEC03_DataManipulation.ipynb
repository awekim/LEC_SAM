#### Practice

```python
# Each student currently occupies one row. Explain why this is wide format.
import pandas as pd
survey = pd.DataFrame({
    "student_id": [101, 102, 103],
    "group": ["A", "B", "A"],
    "stress_before": [7, 5, 8],
    "stress_after": [4, 6, 5],
    "sleep_before": [5, 7, 4],
    "sleep_after": [7, 6, 6]
})
survey
```

```python
# Convert the four measurement columns into two columns named measure and value.
# Keep both student_id and group as identifiers.
survey_long = survey.melt(
    id_vars=["student_id", "group"],
    value_vars=[
        "stress_before", "stress_after",
        "sleep_before", "sleep_after"
    ],
    var_name="measure",
    value_name="value"
)
survey_long
```

```python
# Restore the original wide structure.
survey_wide = (
    survey_long
    .pivot(
        index=["student_id", "group"],
        columns="measure",
        values="value"
    )
    .reset_index()
)
survey_wide
```
